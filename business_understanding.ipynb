# Project Overview

1. Business Understanding: State the question, decision, and hypothesis.
2. Data Understanding: Describe each source (size, gaps, shape of the distributions) and evaluate confidence that the data is fit for the job.
3. Data Preparation: Clean, join on country and year, keep 1950 onward, and build log GDP and groups to create one clean table of 10,000+ rows.
4. Modelling: Run correlations, conditional probabilities, and a simple probability model to determine how strong the link is and how the payoff changes with income.
5. Evaluation: Conduct three hypothesis tests to gather evidence that the pattern is not due to chance.
6. Deployment: Build charts in a live Streamlit app, slides, and an article to provide an answer a non-technical reader can use.

How the Hypothesis Will Be Tested

Test 1 (ANOVA) tests the null hypothesis that average life expectancy is the same in all four GDP groups against the alternative hypothesis that at least one GDP group differs. 

Test 2 (Chi-square) tests if GDP group and child mortality group are independent or related.


 Test 3 (A/B-style comparison using Welch t-test) tests whether average life expectancy is the same before 2000 and from 2000 onward, or if the two periods differ. The null hypothesis will be rejected if the p-value is below 0.05. Test 3 compares two natural groups that were not randomly assigned, so it can show a difference, not a cause.

Regarding what the evidence would mean: A strong, steady link at every income, and little difference between countries with the same income suggests growth first is a reasonable strategy. A strong link, but gains that flatten at high income, or big differences at the same income suggests growth helps but is not enough, so direct health investment also matters. A weak or unstable link suggests income is a poor guide, and other factors need more attention.

For correlation (r) descriptions: 0.7 or higher is strong, 0.3 to 0.7 is moderate, and below 0.3 is weak.

# Who This Is For

Health ministries and treasuries, with Kenya as the local case, need to know how much of their health outcomes income is likely to deliver, and how much needs direct action. Development banks and donors need to know which income levels gain most from outside support. County and regional planners need to know where a place sits compared with others at a similar income. The general reader for the article audience needs a clear answer to whether getting richer makes people healthier.

# Scope

Health measures include life expectancy and under-five (child) mortality. The wealth measure is GDP per capita in international dollars, adjusted so a dollar buys the same in every country and year. The period is 1950 onward, because coverage before 1950 is thin and uneven. The case study is Kenya, to make the findings concrete.

# Risks and Limits

Correlation is not causation because wealth may improve health, health may improve wealth, or something else such as education may drive both; the action is to say so plainly and avoid causal wording. Averages hide inequality because a national figure says nothing about differences inside a country; the action is to state this limit in the write-up. The same country appears in many years, meaning rows are not independent so p-values can look stronger than they should; the action is to report effect sizes and run country-level checks. Health spending is not in the data, so it can show how much income leaves unexplained, but not what fills the gap; the action is to describe the gap and not its cause. Groups created are choices since median splits and before or after 2000 were not randomly assigned; the action is to say so with each test. Child mortality unit is not yet confirmed and axis labels need the right unit; the action is to check the Our World in Data page.
